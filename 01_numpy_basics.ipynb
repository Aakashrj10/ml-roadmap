```python
import numpy as np
X = np.random.randn(100, 3)
print(X.shape)
```

```python
col_means = X.mean(axis=0)
print(col_means)
print(col_means.shape)
```

```python
X_centered = X - col_means
print(X_centered.shape)
print(X_centered.mean(axis=0))
```

```python
n = X.shape[0]
cov_manual = X_centered.T @ X_centered / (n-1)
print(cov_manual.shape)
print(cov_manual)
```

```python
cov_numpy = np.cov(X, rowvar=False)
print(np.allclose(cov_manual, cov_numpy))
```

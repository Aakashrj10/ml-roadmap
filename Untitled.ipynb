```python
import numpy as np

np.random.seed(0)
x = np.random.rand(50)
y = 3 * x + 2 + 0.1 * np.random.randn(50)

print(x.shape, y.shape)
print(x[:3])
print(y[:3])
```

```python
w = 0.0
b = 0.0
y_pred = w * x + b
loss = np.mean((y_pred - y) ** 2)
print(y_pred[:3])
print(loss)
```

```python
error = y_pred - y
dw = np.mean(2 * error * x)
db = np.mean(2 * error)

lr = 0.1
w = w - lr * dw
b = b - lr * db

print(dw, db)
print(w, b)
```

```python
y_pred = w * x + b
loss = np.mean((y_pred - y) ** 2)
print(loss)
```

```python
w = 0.0
b = 0.0
lr = 0.1

for step in range(1000):
    y_pred = w * x + b
    error = y_pred - y
    loss = np.mean(error ** 2)

    dw = np.mean(2 * error * x)
    db = np.mean(2 * error)
    w = w - lr * dw
    b = b - lr * db

    if step % 100 == 0 :
        print(step, loss)

print(w, b)
```

```python
def train(lr, steps=1000):
    w, b = 0.0, 0.0
    for step in range(steps):
        y_pred = w * x + b
        error = y_pred - y
        loss = np.mean(error ** 2)
        dw = np.mean(2 * error * x)
        db = np.mean(2 * error)
        w = w - lr * dw
        b = b - lr * db
        if step % 100 == 0:
            print(step, loss)
    print("final:", w, b)

train(0.001)
        
```
